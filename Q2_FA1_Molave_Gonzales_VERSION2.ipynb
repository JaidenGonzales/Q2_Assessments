```python
def add_numbers(num1, num2):
  sum = num1 + num2
  return sum

def subtract_numbers(num1, num2):
  difference = num1 - num2
  return difference

def multiply_numbers(num1, num2):
  product = num1 * num2
  return product

def divide_numbers(num1, num2):
  if num2 == 0:
    print("Cannot divide by zero!")
  else:
    return num1 / num2

#These are the functions I made that makes the code come to life

print("--Modular Calculator--\n")

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
#This asks the user to input the 2 values they want to use for the calculator


print("\nChoose operation:")
print("1 - Addition:")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division\n")

#Displays what the options the user could use for the calculator
operation = int(input("Enter Choice: "))
#This input asks what operation the user would want to use

if operation == 1:
   print("\nthe sum is:", add_numbers(num1, num2))
elif operation == 2:
  print("the difference is:", subtract_numbers(num1, num2))
elif operation == 3:
  print("the product is:", multiply_numbers(num1, num2))
else:
  print("the quotient is:", divide_numbers(num1, num2))
#Checks what operation the user chose and calculates the answer

print("\n-END-")

#REFLECTION
#1. I made 4 functions in my program: add_numbers, subtract_numbers, multiply_numbers, and divide_numbers
#2. The program only had 2 parameters, num1 and num2
#3. The arguments passed were num1 and num2, which contain the numbers entered by the user. These values are passed to the selected function depending on the operation chosen.
#4. The program used the returned value to display the result of the selected operation. The return statement sends the calculated answer back to the function call, and the print() statement displays it to the user.
#5. Dividing the program into functions makes the code more organized, readable, and easier to understand. It also makes debugging and updating the program easier because each function has a specific task.
```
